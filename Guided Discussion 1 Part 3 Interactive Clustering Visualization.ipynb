```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

X, y_true = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=1.2,
    random_state=42
)

X = StandardScaler().fit_transform(X)
```

```python
from sklearn.cluster import KMeans, DBSCAN
def kmeans_visualization(k=4):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(X)
    
    plt.figure(figsize=(8,6))
    plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis', s=50)
    
    # Plot cluster centers
    centers = model.cluster_centers_
    plt.scatter(
        centers[:,0],
        centers[:,1],
        c='red',
        s=250,
        marker='X',
        label='Centers'
    )
    
    plt.title(f"K-Means Clustering (k={k})")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.legend()
    plt.show()
```

```python
import ipywidgets as widgets
from ipywidgets import interact

interact(
    kmeans_visualization,
    k=widgets.IntSlider(
        value=4,
        min=2,
        max=10,
        step=1,
        description='Clusters (k)'
    )
);
```

```python
def dbscan_visualization(eps=0.5, min_samples=5):
    model = DBSCAN(eps=eps, min_samples=min_samples)
    labels = model.fit_predict(X)
    
    plt.figure(figsize=(8,6))
    
    unique_labels = set(labels)
    colors = plt.cm.viridis(np.linspace(0, 1, len(unique_labels)))
    
    for label, color in zip(unique_labels, colors):
        if label == -1:
            color = 'black'
            label_name = "Noise"
        else:
            label_name = f"Cluster {label}"
        
        plt.scatter(
            X[labels == label, 0],
            X[labels == label, 1],
            c=[color],
            s=50,
            label=label_name
        )
    
    plt.title(f"DBSCAN (eps={eps}, min_samples={min_samples})")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.legend()
    plt.show()
```

```python
interact(
    dbscan_visualization,
    eps=widgets.FloatSlider(
        value=0.5,
        min=0.1,
        max=2.0,
        step=0.1,
        description='Epsilon'
    ),
    min_samples=widgets.IntSlider(
        value=5,
        min=2,
        max=20,
        step=1,
        description='Min Samples'
    )
);
```

# Reflection Questions

What happens when algorithm parameters change?
The clusters change when the parameters change. When Epsilon is lowered, there are more clusters and more noise points. Obviously, when K-increases, the number of clusters increase. The min samples parameters does not change much with this since the clusters are so distinct.

Does clustering improve or worsen with certain parameter choices?
Yes, there are specific values where the clustering looks best.

In what situations would this algorithm fail?
When the clusters are not as distinct and when there are not clear centroids. It could have a lot of noise points if the clusters are not dense and distinct.

How do results compare to K-Means on the same dataset?
Results are similar for the blobs data because the clusters are distinct and both algorithms perform well. DBScan can get the same amount of clusters with certain epsilon values as k-means with k equal to 3 or 4.
