```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', as_frame=False)
```

```python
X, y = mnist.data, mnist.target
```

```python
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]
```

```python
some_digit = X[7]
print(some_digit)
```

```python
import numpy as np
def shift_image(input_image, direction):  
    array_784 = np.arange(784) 

    # Reshape the 1D array to a 2D array of 28x28
    image = input_image.reshape(28, 28)
    
    shifted_image = np.zeros(image.shape, dtype=image.dtype)
    if direction == 'left':
        shifted_image[:, :-1] = image[:, 1:]
    if direction == 'right':
        shifted_image[:, 1:] = image[:, :-1]
    if direction == 'up':
        shifted_image[-1:, :] = image[:1, :]
    if direction == 'down':
        shifted_image[1:, :] = image[:-1, :]
    
    output_image = shifted_image.reshape(784)
    return output_image
```

```python
random_shifted = shift_image(some_digit, 'left')
print(random_shifted)
```

```python
shifted_left = []
for image in X_train:
    shifted_left.append(shift_image(image, 'left'))
    
shifted_right = []
for image in X_train:
    shifted_right.append(shift_image(image, 'right'))
    
shifted_up = []
for image in X_train:
    shifted_up.append(shift_image(image, 'up'))
    
shifted_down = []
for image in X_train:
    shifted_down.append(shift_image(image, 'down'))
```

```python
X_train_new = X_train + np.array(shifted_left) + np.array(shifted_right) + np.array(shifted_up) + np.array(shifted_down)
```

```python
from sklearn.linear_model import SGDClassifier

sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train)
```

```python
from sklearn.model_selection import cross_val_score

cross_val_score(sgd_clf, X_train, y_train, cv=3, scoring="accuracy")
```
